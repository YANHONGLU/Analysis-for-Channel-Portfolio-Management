#1.Transaction Monitoring Rule

Identify suspicious customers: If a customer's total transaction amount exceeds £10,000 in a day, or if they have more than 3 high-risk transactions in a day, they are flagged.

```python
import pandas as pd
```

```python
# Group by customer and date → Calculate total amount → Calculate number of high-risk transactions → If rules are exceeded...flag 'transactions' table:  customer_id,date,amount,high_risk
def flag_suspicious(transactions):
    summary = transactions.groupby(["customer_id", "date"]).agg(total_amount=("amount", "sum"),high_risk_count=("high_risk", "sum"))
    suspicious = summary[(summary["total_amount"] > 10000) | (summary["high_risk_count"] >= 3)    ]
    return suspicious
```

#2.Model Performance Monitoring

Last month's model AUC was 0.80. Calculate this month's AUC; if it decreases by more than 0.05, an alert will be triggered.

```python
from sklearn.metrics import roc_auc_score
```

```python
def check_model(y_true, probabilities, previous_auc):
 # roc_auc_score = To see how accurate the model's guesses are, and whether it can distinguish between "yes" and "no"
    current_auc = roc_auc_score(y_true,probabilities ) # y_true real result ;probabilities model's guessed answer/probability
    performance_drop = previous_auc - current_auc
    alert = performance_drop > 0.05
    return current_auc, alert
```
